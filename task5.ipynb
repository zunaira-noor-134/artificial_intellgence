```python
menu = {
        1: ("Burger", 500),
        2: ("Pizza", 800),
        3: ("Fries", 250),
        4: ("Drink", 150)
       }

orders = []

while True:
    print("\n1. Burger Rs.500")
    print("2. Pizza Rs.800")
    print("3. Fries Rs.250")
    print("4. Drink Rs.150")
    
    choice = int(input("Enter item number: "))
    qty = int(input("Enter quantity: "))
    
    name = menu[choice][0]
    price = menu[choice][1]
    
    orders.append((name, price, qty))
    
    more = input("Order more? (yes/no): ")
    if more == "no":
        break

total = 0
print("bill")
for name, price, qty in orders:
    item_total = price * qty
    total = total + item_total
    print(f"{name} x {qty} = Rs.{item_total}")

print("Total Bill: Rs.", total)

f = open("orders.txt", "w")
f.write("bill")
for name, price, qty in orders:
    f.write(f"{name} x {qty} = Rs.{price * qty}\n")
f.write(f"Total: Rs.{total}")
f.close()

print("Saved to orders.txt")
```
