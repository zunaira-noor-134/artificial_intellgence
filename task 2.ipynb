```python
sentence = input("Enter a sentence")
print(sentence)
words = sentence.split()
count = len(sentence)
print(words)
print(count)
```

```python
largest = words[0]              
for w in words:                 
    if len(w) > len(largest):  
        largest = w            
print("Largest:",largest)
```

```python
shortest = words[0]              
for w in words:                 
    if len(w) < len(shortest):  
        shortest = w            
print("shortest:",shortest)
```

```python
vowel = 0
for char in sentence:
    if char.lower() in "aeiou":
        vowel = vowel + 1

print("Number of vowels:", vowel)
```

```python
unique_words = set(words)
print("Unique words:", unique_words)
```

```python
sorted_words = sorted(words)
print(sorted_words)
```

```python
with open("word_analysis.txt", "w") as f:
    f.write(f"Sentence: {sentence}\n")
    f.write(f"Words: {words}\n")
    f.write(f"Characters: {len(sentence)}\n")
    f.write(f"Largest: {max(words, key=len)}\n")
    f.write(f"Shortest: {min(words, key=len)}\n")
    f.write(f"Vowels: {sum(1 for c in sentence if c.lower() in 'aeiou')}\n")
    f.write(f"Unique: {set(words)}\n")
    f.write(f"Sorted: {sorted(words)}\n")

print("File saved!")
```

```python
def analyze_sentence(sentence):
    words = sentence.split()
    
    largest = words[0]
    for w in words:
        if len(w) > len(largest):
            largest = w
    
    shortest = words[0]
    for w in words:
        if len(w) < len(shortest):
            shortest = w
    
    vowel_count = 0
    for char in sentence:
        if char.lower() in "aeiou":
            vowel_count = vowel_count + 1
    
    unique_words = set(words)
    sorted_words = sorted(words)
    
    return words, len(sentence), largest, shortest, vowel_count, unique_words, sorted_words


sentence = input("Enter a sentence: ")

words, char_count, largest, shortest, vowel_count, unique_words, sorted_words = analyze_sentence(sentence)

with open("word_analysis.txt", "w") as f:
    f.write(f"Sentence: {sentence}\n")
    f.write(f"Words: {words}\n")
    f.write(f"Characters: {char_count}\n")
    f.write(f"Largest word: {largest}\n")
    f.write(f"Shortest word: {shortest}\n")
    f.write(f"Vowels: {vowel_count}\n")
    f.write(f"Unique words: {unique_words}\n")
    f.write(f"Alphabetical order: {sorted_words}\n")

print("Analysis saved to word_analysis.txt")
```
