```python
students = []

for i in range(5):
    name = (input("Enter name: "))
    rollno = input("Enter rollno: ")
    python_marks=int(input("enter python marks"))
    ai_marks=int(input("enter ai marks"))
    math_marks=int(input("enter math marks"))

    student={
    "name": name,
    "rollno":rollno,
    "python_marks":python_marks,
    "ai_marks":ai_marks,
    "math_marks":math_marks
    }
    students.append(student)
print(students)
```

```python
def cal_results(student):
    total_marks= student["python_marks"] + student["ai_marks"]+ student["math_marks"]
    percentage= total_marks/3 
    if percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "F"
    
    return total_marks, percentage, grade 
print("RESULTS")

for s in students:
    total, percentage, grade = cal_results(s)
    print(f"Name: {s['name']}, Roll: {s['rollno']}, Total: {total}, Percentage: {percentage}%, Grade: {grade}")
    
```

```python
highest = students[0]
for s in students:
    if cal_results(s)[1] > cal_results(highest)[1]:
        highest = s
total, percentage, grade = cal_results(highest)
print("highest percentage")
print(f"Name: {highest['name']}, Roll: {highest['rollno']}, Percentage: {percentage}%, Grade: {grade}")
```

```python
f = open("results.txt", "w")
for s in students:
    total, percentage, grade = cal_results(s)
    f.write(f"Name: {s['name']}, Roll: {s['rollno']}, Total: {total}, Percentage: {percentage:.1f}%, Grade: {grade}\n")
total, percentage, grade = cal_results(highest)
f.write(f"Top Student: {highest['name']}, Percentage: {percentage:.1f}%, Grade: {grade}\n")
f.close()
print("File saved!")
```
