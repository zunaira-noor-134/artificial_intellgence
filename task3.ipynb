```python
def calculate_bill(products):
    bill = 0
    for p in products:
        bill = bill + (p["price"] * p["quantity"])
    if bill > 10000:
        discount = bill * 0.10
        final_bill = bill - discount
    else:
        discount = 0
        final_bill = bill
    return bill, discount, final_bill


products = []

for i in range(5):
    name = input("Enter product name: ")
    price = float(input("Enter price: "))
    quantity = int(input("Enter quantity: "))
    
    product = {
        "name": name,
        "price": price,
        "quantity": quantity
    }
    products.append(product)

bill, discount, final_bill = calculate_bill(products)

most_expensive = products[0]
for p in products:
    if p["price"] > most_expensive["price"]:
        most_expensive = p

print("finall bill")
for p in products:
    print(f"{p['name']} x {p['quantity']} = Rs. {p['price'] * p['quantity']}")

print(f"\nTotal Bill: Rs. {bill}")
print(f"Discount: Rs. {discount}")
print(f"Final Bill: Rs. {final_bill}")
print(f"\nMost Expensive: {most_expensive['name']} (Rs. {most_expensive['price']})")

with open("bill.txt", "w") as f:
    f.write("shopping bill")
    for p in products:
        f.write(f"{p['name']} x {p['quantity']} = Rs. {p['price'] * p['quantity']}\n")
    f.write(f"\nTotal Bill: Rs. {bill}\n")
    f.write(f"Discount: Rs. {discount}\n")
    f.write(f"Final Bill: Rs. {final_bill}\n")
    f.write(f"\nMost Expensive: {most_expensive['name']}\n")

print("saved to bill.txt")
```
